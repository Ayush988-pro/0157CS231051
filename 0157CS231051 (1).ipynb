```python
##Quiz App with FILE HANDLING
   # 1. registration
   # 2. login (user and admin)
    #3. quiz
       ## 3.1 login
        3.2 attempt quiz
            3.2.a category (DSA, DBMS, PYTHON)
                display atleast 5-10 question randomly(shuffle)
                at a time 1 question with options
        3.3 score
            enrollment    category    marks/total marks   datetime
        3.4 logout
        3.5 update profile, email, branch, year
            contact, name,
        3.6 profile
    4. exit"


def register():
    users = load_data("users.txt")
    enroll = input("Enter enrollment number: ")
    if enroll in users:
        print("Already registered")
        return
    users[enroll] = [
        input("Enter password: "),
        input("Enter name: "),
        input("Enter email: "),
        input("Enter branch: "),
        input("Enter year: "),
        input("Enter contact: ")
    ]
    save_data("users.txt", users)
    print("Registered successfully")

def login():
    users = load_data("users.txt")
    enroll = input("Enter enrollment: ")
    pwd = input("Enter password: ")
    if enroll in users and users[enroll][0] == pwd:
        print("Login successful")
        user_menu(enroll)
    elif enroll == "admin" and pwd == "admin":
        print("Admin login successful")
        admin_menu()
    else:
        print("Invalid credentials")

def user_menu(enroll):
    while True:
        print("\n1.Attempt Quiz 2.View Score 3.Profile 4.Update Profile 5.Logout")
        ch = input("Enter choice: ")
        if ch == "1":
            attempt_quiz(enroll)
        elif ch == "2":
            show_score(enroll)
        elif ch == "3":
            profile(enroll)
        elif ch == "4":
            update_profile(enroll)
        elif ch == "5":
            break

def admin_menu():
    try:
        with open("scores.txt", "r") as f:
            print(f.read())
    except:
        print("No scores yet")

def attempt_quiz(enroll):
    print("Select Category: 1.DSA 2.DBMS 3.PYTHON")
    c = input("Enter choice: ")
    if c == "1":
        q, options, ans = "What is a stack?", ["LIFO", "FIFO", "Tree", "Graph"], "1"
        category = "DSA"
    elif c == "2":
        q, options, ans = "What is a primary key?", ["Unique key", "Duplicate key", "Foreign key", "None"], "1"
        category = "DBMS"
    else:
        q, options, ans = "What is the output of print(2**3)?", ["6", "8", "9", "5"], "2"
        category = "PYTHON"
    print("\n", q)
    for i in range(len(options)):
        print(i + 1, options[i])
    user_ans = input("Enter option number: ")
    score = 1 if user_ans == ans else 0
    with open("scores.txt", "a") as f:
        f.write(f"{enroll},{category},{score}/1\n")
    print("Your score:", score, "/1")

def show_score(enroll):
    try:
        with open("scores.txt", "r") as f:
            found = False
            for line in f:
                if line.startswith(enroll):
                    print(line.strip())
                    found = True
            if not found:
                print("No records found")
    except:
        print("No records found")

def profile(enroll):
    users = load_data("users.txt")
    print("Name:", users[enroll][1])
    print("Email:", users[enroll][2])
    print("Branch:", users[enroll][3])
    print("Year:", users[enroll][4])
    print("Contact:", users[enroll][5])

def update_profile(enroll):
    users = load_data("users.txt")
    data = users[enroll]
    data[2] = input("Enter new email: ")
    data[3] = input("Enter new branch: ")
    data[4] = input("Enter new year: ")
    data[5] = input("Enter new contact: ")
    data[1] = input("Enter new name: ")
    users[enroll] = data
    save_data("users.txt", users)
    print("Profile updated")

def load_data(fname):
    data = {}
    try:
        with open(fname, "r") as f:
            for line in f:
                parts = line.strip().split("|")
                data[parts[0]] = parts[1:]
    except:
        pass
    return data

def save_data(fname, data):
    with open(fname, "w") as f:
        for k, v in data.items():
            f.write(k + "|" + "|".join(v) + "\n")

def main():
    while True:
        print("\n1.Register 2.Login 3.Exit")
        ch = input("Enter choice: ")
        if ch == "1":
            register()
        elif ch == "2":
            login()
        elif ch == "3":
            break

main()
```
