```python
students = {}
current_user = None

def register():
    username = input("Username: ")
    if username in students:
        print("Username already exists")
        return
    password = input("Password: ")
    name = input("Full Name: ")
    age = input("Age: ")
    gender = input("Gender: ")
    email = input("Email: ")
    phone = input("Phone: ")
    city = input("City: ")
    state = input("State: ")
    course = input("Course: ")

    students[username] = {
        "password": password,
        "name": name,
        "age": age,
        "gender": gender,
        "email": email,
        "phone": phone,
        "city": city,
        "state": state,
        "course": course
    }

    print("Registration Successful")

def login():
    global current_user
    username = input("Username: ")
    password = input("Password: ")
    if username in students and students[username]["password"] == password:
        current_user = username
        print("Login Successful")
    else:
        print("Invalid Login")

def show_profile():
    if not current_user:
        print("Login First")
        return
    data = students[current_user]
    for k, v in data.items():
        if k != "password":
            print(k, ":", v)

def update_profile():
    if not current_user:
        print("Login First")
        return
    data = students[current_user]
    for k in data:
        if k != "password":
            print(k, ":", data[k])
    field = input("Enter field to update: ")
    if field in data and field != "password":
        data[field] = input("Enter new value: ")
        print("Updated")
    else:
        print("Invalid field")

def logout():
    global current_user
    current_user = None
    print("Logged out")

while True:
    print("\n1.Register\n2.Login\n3.Show Profile\n4.Update Profile\n5.Logout\n6.Exit")
    ch = input("Choose: ")
    if ch == "1":
        register()
    elif ch == "2":
        login()
    elif ch == "3":
        show_profile()
    elif ch == "4":
        update_profile()
    elif ch == "5":
        logout()
    elif ch == "6":
        break
    else:
        print("Invalid")
```
